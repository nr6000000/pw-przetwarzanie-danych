```python
import pandas as pd

train_df = pd.read_csv('ag_news_csv/train.csv', names=['class', 'title', 'description'])
test_df = pd.read_csv('ag_news_csv/test.csv', names=['class', 'title', 'description'])
train_df.head()
```

```python
import re

import matplotlib.pyplot as plt
from wordcloud import STOPWORDS, WordCloud


def generate_wordcloud(text):
    text = re.sub(r'[^A-Za-z\s]', '', text) 

    text = text.lower()

    stopwords = set(STOPWORDS)
    text = ' '.join(word for word in text.split() if word not in stopwords)

    wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')  
    plt.show()

```

```python
text = ' '.join(train_df['title'].astype(str).tolist())
text += ' ' + ' '.join(train_df['description'].astype(str).tolist())
generate_wordcloud(text)
```

```python
df = train_df[train_df['class'] == 1]
text = ' '.join(df['title'].astype(str).tolist())
text += ' ' + ' '.join(df['description'].astype(str).tolist())
generate_wordcloud(text)
```
